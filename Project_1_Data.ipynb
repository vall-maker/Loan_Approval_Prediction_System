# Data Simulation: Population & Baseline Profiles

The first stage of the project involves synthesizing a realistic dataset of 3,000 loan applicants. To ensure the model learns meaningful relationships, the data is not purely random but follows conditional logic based on employment status.

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
n = 3000
```

# Setup and Reproducibility

 - Libraries: Numpy is utilized for random number generation, and Pandas is used for structured data storage.

 - Seeding: np.random.seed(42) is implemented to ensure that the synthetic dataset remains consistent across every execution of the notebook.

 - Sample Size: The variable n = 3000 defines the population size for the simulation.

```python
user_age = np.random.randint(21,60,n)
```

```python
loan_amount = np.random.randint(2000,1000000,n)
```

# Independent Feature Generation

Initial variables are generated using discrete uniform distributions to represent a diverse demographic:

 - User Age: Applicants are aged between 21 and 60 years.

 - Loan Amount: Requested loan values range from $2,000 to $1,000,000.

```python
income = []
```

```python
credit_score = []
```

```python
employment_status = np.random.choice(["Employed","Unemployed","Self-employed"], n)
for status in employment_status:
    if status == "Employed":
        inc = np.random.randint(30000,1000000)
        cs = np.random.randint(500,850)
    elif status == "Unemployed":
        inc = np.random.randint(5000,40000)
        cs = np.random.randint(300,500)
    else:
        inc = np.random.randint(5000,1000000)
        cs = np.random.randint(300,850)
    # np.append(income,inc)
    income.append(inc)
    credit_score.append(cs)
```

# Conditional Profile Logic

To mimic real-world financial correlations, Income and Credit Score are generated based on an applicant's Employment Status

# Data Simulation: Financial Metrics & Final Assembly
The second stage of the simulation focuses on generating debt-related features, historical behavior, and consolidating all individual arrays into a structured dataset.

```python
loan_term = np.random.randint(5,24,n)
```

```python
existing_debt = np.random.randint(1000, 400000, n)
```

```python
loan_to_income_ratio = existing_debt/income
```

# Debt and Ratio Calculations

To provide the model with indicators of financial burden, the following variables are generated:

 - Loan Term: The duration of the requested loan is set between 5 and 24 months.

 - Existing Debt: A random value between $1,000 and $400,000 is assigned to represent the applicant's current liabilities.

 - Loan-to-Income Ratio: This critical feature is calculated by dividing existing_debt by income, allowing the model to assess debt relative to earning capacity.

```python
payment_history = np.random.choice(["Good","Bad","Average"],n)
```

```python
num_of_defaults = []
for payment in payment_history:
    if payment == "Good":
        nod = np.random.randint(0,3)
    elif payment == "Average":
        nod = np.random.randint(3,6)
    else:
        nod = np.random.randint(6,10)
    num_of_defaults.append(nod)
```

# Behavioral Simulation: Payment History & Defaults

To simulate creditworthiness over time, the code uses conditional logic based on a categorized payment history:

 - Payment History: Applicants are randomly assigned a status of "Good", "Average", or "Bad".

 - Number of Defaults: The specific count of defaults (num_of_defaults) is then derived from that history:

 - Good: 0 to 2 defaults.

 - Average: 3 to 5 defaults.

 - Bad: 6 to 9 defaults.

```python
savings_balance = []
for i in income:
    sb = (30/100)*i
    savings_balance.append(sb)
```

# Asset Simulation: 

Savings BalanceA simplified savings model is implemented where the savings_balance is calculated as a fixed 30% of the applicant's income ($sb = (30/100) * i$).

```python
df = pd.DataFrame({
    'user_age': user_age,
    'income': income,
    'employment_status': employment_status,
    'credit_score': credit_score,
    'loan_amount': loan_amount,
    'loan_term': loan_term,
    'existing_debt': existing_debt,
    'loan_to_income_ratio': loan_to_income_ratio,
    'payment_history': payment_history,
    'num_of_defaults': num_of_defaults,
    'savings_balance': savings_balance
})
```

```python
df.to_csv('loan_data.csv', index=False)
```
